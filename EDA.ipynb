```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file_path = "Agent_KIO_Dataset_FROZEN.xlsx"
df = pd.read_excel(file_path, sheet_name=0)
```

```python
display(df.head())
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
print(" Duplicates Check")
print("="*50)
print(f"- Exact duplicate rows: {df.duplicated().sum()}")
print(f"- Duplicate text messages (Message): {df['Message'].duplicated().sum()}")
print(f"- Duplicate IDs (ID): {df['ID'].duplicated().sum()}")
```

```python
for col in ['Category', 'Flag', 'Split', 'Type']:
    print(f"--- {col} Value Counts ---")
    print(df[col].value_counts())
    print("\n")
```

```python

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

sns.countplot(data=df, x='Category', ax=axes[0, 0], palette='Set2')
axes[0, 0].set_title('Distribution of Category')

sns.countplot(data=df, x='Flag', ax=axes[0, 1], palette='Set2')
axes[0, 1].set_title('Distribution of Flag')

sns.countplot(data=df, x='Split', ax=axes[1, 0], palette='Set2')
axes[1, 0].set_title('Distribution of Split (Train/Test)')

sns.countplot(data=df, x='Type', ax=axes[1, 1], palette='Set2')
axes[1, 1].set_title('Distribution of Message Type')

plt.tight_layout()
plt.show()
```
