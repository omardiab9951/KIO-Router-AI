```python
import pandas as pd
from google.colab import files
```

```python
uploaded = files.upload()
filename = next(iter(uploaded))
```

```python
df = pd.read_excel(filename)
```

```python
train_df = df[df["Split"] == "Train"].copy()
test_df = df[df["Split"] == "Test"].copy()
```

```python
print(pd.crosstab(df["Split"], df["Category"]))
```

```python
assert len(df) == 60
assert df["Category"].value_counts().to_dict() == {
    "Support": 30, "Sales": 30
}
assert train_df["Category"].value_counts().to_dict() == {
    "Support": 20, "Sales": 20
}
assert test_df["Category"].value_counts().to_dict() == {
    "Support": 10, "Sales": 10
}
assert df["Flag"].eq(df["Category"].map({
    "Support": 0, "Sales": 1
})).all()
assert df["Message"].notna().all()
assert not df["Message"].str.strip().str.lower().duplicated().any()

print("\nDataset checks passed!")
```
