```python
from pathlib import Path
import pandas as pd

DATA_FILENAME = 'Agent_KIO_Dataset_FROZEN.xlsx'
candidates = [Path.cwd() / DATA_FILENAME] + [p / DATA_FILENAME for p in Path.cwd().parents]
data_path = next((p for p in candidates if p.is_file()), None)
if data_path is None:
    matches = list(Path.cwd().glob(f'**/{DATA_FILENAME}'))
    data_path = matches[0] if matches else None
if data_path is None:
    raise FileNotFoundError(f'Could not find {DATA_FILENAME}. Open the notebook from the System folder or update DATA_FILENAME.')
print(f'Loading dataset from: {data_path.resolve()}')
df = pd.read_excel(data_path, engine='openpyxl')
df.head()
```

```python
train_df = df[df["Split"] == "Train"].copy()
test_df = df[df["Split"] == "Test"].copy()
```

```python
assert len(df) == 60
assert df["Category"].value_counts().to_dict() == {
    "Support": 30, "Sales": 30
}
assert train_df["Category"].value_counts().to_dict() == {
    "Support": 20, "Sales": 20
}
assert test_df["Category"].value_counts().to_dict() == {
    "Support": 10, "Sales": 10
}
assert df["Flag"].eq(df["Category"].map({
    "Support": 0, "Sales": 1
})).all()
assert df["Message"].notna().all()
assert not df["Message"].str.strip().str.lower().duplicated().any()

print("\nDataset checks passed!")
```
